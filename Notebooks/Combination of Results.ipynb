```python
# Importing the CSV files 

# Baseline
abstract = "Baseline"
dataset = "Enron"
Baseline_enron_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
Baseline_enron_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
Baseline_enron_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

dataset = "LingSpam"
Baseline_lingSpam_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
Baseline_lingSpam_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
Baseline_lingSpam_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

dataset = "SpamAssassin"
Baseline_spamAssassin_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
Baseline_spamAssassin_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
Baseline_spamAssassin_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

# Cleaned
abstract = "Cleaned"
dataset = "Enron"
Cleaned_enron_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
Cleaned_enron_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
Cleaned_enron_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

dataset = "LingSpam"
Cleaned_lingSpam_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
Cleaned_lingSpam_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
Cleaned_lingSpam_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

dataset = "SpamAssassin"
Cleaned_spamAssassin_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
Cleaned_spamAssassin_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
Cleaned_spamAssassin_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

# NER
abstract = "NER"
dataset = "Enron"
NER_enron_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
NER_enron_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
NER_enron_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

dataset = "LingSpam"
NER_lingSpam_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
NER_lingSpam_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
NER_lingSpam_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')

dataset = "SpamAssassin"
NER_spamAssassin_ml = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_machinelearning.csv')
NER_spamAssassin_dl = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_deeplearning.csv')
NER_spamAssassin_tr = pd.read_csv(f'G:/My Drive/Dissertation/Results/{abstract}/{dataset}/Results/result_transformers.csv')
```

```python
# Combining the CSV files into abstracts and shortening the model names for the table
# Update the names with the corresponding preprocessing transformation, 
# such as clean data to +C and Entity-rule to +E 

Enron = pd.concat([
    Baseline_enron_ml, Baseline_enron_dl, Baseline_enron_tr,
    Cleaned_enron_ml, Cleaned_enron_dl, Cleaned_enron_tr,
    NER_enron_ml, NER_enron_dl, NER_enron_tr,
], axis=0)

Enron["Model"] = Enron["Model"].replace({
    "Multinomial Naive Bayes": "MNB",
    "Logistic Regression": "LR",
    "Linear SVM": "LSVM",
    "DistilBERT": "DBERT",
    "BiLSTM": "BLSTM"
})

Enron.iloc[7:14, 0] = Enron.iloc[7:14, 0] + " (+C)"
Enron.iloc[14:, 0] = Enron.iloc[14:, 0] + " (+E)"

LingSpam = pd.concat([
    Baseline_lingSpam_ml, Baseline_lingSpam_dl, Baseline_lingSpam_tr,
    Cleaned_lingSpam_ml, Cleaned_lingSpam_dl, Cleaned_lingSpam_tr,
    NER_lingSpam_ml, NER_lingSpam_dl, NER_lingSpam_tr,
], axis=0)

LingSpam["Model"] = LingSpam["Model"].replace({
    "Multinomial Naive Bayes": "MNB",
    "Logistic Regression": "LR",
    "Linear SVM": "LSVM",
    "DistilBERT": "DBERT",
    "BiLSTM": "BLSTM"
})

LingSpam.iloc[7:14, 0] = LingSpam.iloc[7:14, 0] + " (+C)"
LingSpam.iloc[14:, 0] = LingSpam.iloc[14:, 0] + " (+E)"

SpamAssassin = pd.concat([
    Baseline_spamAssassin_ml, Baseline_spamAssassin_dl, Baseline_spamAssassin_tr,
    Cleaned_spamAssassin_ml, Cleaned_spamAssassin_dl, Cleaned_spamAssassin_tr,
    NER_spamAssassin_ml, NER_spamAssassin_dl, NER_spamAssassin_tr
],axis=0)

SpamAssassin["Model"] = SpamAssassin["Model"].replace({
    "Multinomial Naive Bayes": "MNB",
    "Logistic Regression": "LR",
    "Linear SVM": "LSVM",
    "DistilBERT": "DBERT",
    "BiLSTM": "BLSTM"
})

SpamAssassin.iloc[7:14, 0] = SpamAssassin.iloc[7:14, 0] + " (+C)"
SpamAssassin.iloc[14:, 0] = SpamAssassin.iloc[14:, 0] + " (+E)"
```

```python
# Remove unnecessary columns from each dataframe.

Enron.drop(['Parameter', 'Time', 'AUC', 'ROC-AUC', 'TP', 'TN', 'FP', 'FN'], axis=1, inplace=True)
LingSpam.drop(['Parameter', 'Time', 'AUC', 'ROC-AUC', 'TP', 'TN', 'FP', 'FN'], axis=1, inplace=True)
SpamAssassin.drop(['Parameter', 'Time', 'AUC', 'ROC-AUC', 'TP', 'TN', 'FP', 'FN'], axis=1, inplace=True)
```

```python
# Round the values from each column to 3 decimal places.

col = ["Accuracy", "Precision", "Recall", "F1-Score", "PR-AUC", "Threshold"]
Enron[col] = Enron[col].round(3)
SpamAssassin[col] = SpamAssassin[col].round(3)
LingSpam[col] = LingSpam[col].round(3)
```

```python
# Saving the results metrics

Enron.to_csv(f'G:/My Drive/Dissertation/Results/Enron_Metrics_Final.csv')
SpamAssassin.to_csv(f'G:/My Drive/Dissertation/Results/SpamAssassin_Metrics_Final.csv')
LingSpam.to_csv(f'G:/My Drive/Dissertation/Results/LingSpam_Metrics_Final.csv')
```
